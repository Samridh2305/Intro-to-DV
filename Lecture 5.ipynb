This notebook walks through an example of deeper Matplotlib customizations to create a compelling visualization.

We will re-create the Marimekko chart (variable width bar plot) from the Reuters article, "Journey into sleep: An illustrated tour of why scientists are finding that sleep matters more than we know."

Article: https://www.reuters.com/graphics/HEALTH-SLEEP/mopakyjmnpa/

```python
import pandas as pd
import matplotlib.pyplot as plt

import math
```

# Load Dataset

Our data comes from the National Sleep Foundation in the USA from a panel comprised of members and sleep experts from various American medical associations.

Specifically, it comes from the following associated published research article:
- "National Sleep Foundation’s sleep time duration recommendations: methodology and results summary," Sleep Health, https://doi.org/10.1016/j.sleh.2014.12.010

```python
df = pd.read_csv('sleep.csv')
```

```python
df
```

We need to clean and preprocess this data up a bit in prep for our visualization.

First, the data set breaks "Young adults" and "Adults" into two different categories. We'll merge these together since that is how the visualization does it.

```python
df.loc[6, 'Max age'] = 65
df.loc[8, 'Max age'] = 75
```

Next, we need to define some limits for what will become the bars in our visualization. We will just store these in our DataFrame for convenience.

They all extend on the y-axis from 0 to 24 hours. We will add columns for that.

We will also set the x-axis positioning of our bars to be the midpoint of each age bracket, and define the width of each bar (we have variable width bars, as a reminder).                    

```python
df['start'] = 0
df['end'] = 24
```

```python
df['x_pos'] = (df['Max age'] + df['Min age']) / 2
df['x_width'] = df['Max age'] - df['Min age']
```

Let's restrict our DataFrame to the essentials of what we need, and rename the columns for convenience.

```python
df = df[['x_pos', 'x_width', 'start', 'Recommended lower', 'Recommended upper', 'end']]
```

```python
df.columns = ['x_pos', 'x_width', 'start', 'lower', 'upper', 'end']
```

```python
df
```

Let's make our plot!

# Visualization Step 1 - Base Image

We can grab the colours used in the original image using most standard image editors. The colours are:
- `#051b2c`
- `#4789be`
- `#0c3759`
- `#f3e2c0`
- `#6ebeec`
- `#95cef2`

We will start by making a variable width bar plot as our base.

```python
fig, ax = plt.subplots(figsize=(15.66/1.5,9/1.5))

for index, row in df.iterrows():

    ax.bar(row['x_pos'], width=row['x_width'], 
           bottom=row['start'], height=row['lower'] - row['start'],  
           color="#051b2c", edgecolor="#4789be")
    
    ax.bar(row['x_pos'], width=row['x_width'], 
           bottom=row['lower'], height=row['upper'] - row['lower'],
           hatch="\\\\\\", color="#4789be")
```

Note we draw two bars for each category. One is the bottom solid bar, and the second is the top hatched bar. 

We can specify the start and end point of the bars along the y-axis using the `bottom=` and `height=` arguments. The `width=` argument sets the individual width of each bar.

Also note the `hatch=` argument for the top bar, and the use of `edgecolor=` to create the separating line between the bottom bars.

# Visualization Step 2 - Background, x/y Axes

Let's fix up the background colours, x/y limits, and annotations around the x/y axes.

```python
fig, ax = plt.subplots(figsize=(15.66/1.5,9/1.5))

for index, row in df.iterrows():

    ax.bar(row['x_pos'], width=row['x_width'], 
           bottom=row['start'], height=row['lower'] - row['start'],  
           color="#051b2c", edgecolor="#4789be")
    
    ax.bar(row['x_pos'], width=row['x_width'], 
           bottom=row['lower'], height=row['upper'] - row['lower'],
           hatch="///////", color="#4789be")
    

# # Set background colour outside the axis
fig.patch.set_facecolor("#0c3759") 

# # Set background colour inside the axis
ax.set_facecolor("#4789be")

ax.set_xlim((0, 75))
ax.set_ylim((0, 24))

ax.set_ylabel('Hours\nof sleep', rotation='horizontal', color='#f3e2c0', fontweight='bold')
ax.yaxis.set_label_coords(-0.07, 0.58)

ax.set_xticks([1, 6, 12, 18, 65])
ax.set_xticklabels(['1', '6', '12', '18', ''])
ax.set_yticks(range(0, 25))
ax.set_yticklabels([0] + ['']*5 + [6] + [''] + [8] + [''] + [10] + [''] + [12] + ['']*5 + [18] + ['']*5 + [24])
ax.tick_params('y', color='#f3e2c0', labelcolor='#f3e2c0')
ax.tick_params('x', length=0, labelcolor='#f3e2c0')

ax.text(-1.5, -1.6, 'year old', color='#f3e2c0')
ax.text(65, -0.85, '65 and older', color='#f3e2c0')

# # Add space between the y-axis tick marks and the frame of the axis
ax.spines['left'].set_position(('outward', 8))
ax.spines['left'].set_visible(False)
```

Several small tweaks made here.

One, the tick mark lengths on the x-axis were set to 0. We won't labels at these locations, but not the tick marks themselves.

Two, the x-axis tick mark labels are centred on the space between bars by default for Matplotlib. This is fine for the 1, 6, 12 and 18 labels, but not for the "65 and older" label, which is right-justified in its horizontal alignment. We can address that by manually adding this text in the correct location.

Three, the y-axis label, "Hours of sleep" is repositioned to make it not perfectly centred along the y-axis, along with the rotation to make it horizontal.

Four, the y-axis tick marks are offset from the frame of the visualization. We can accomplish this by repositioning the left spine outward slightly.

How do we make the sun and moon icons? 

Well, the sun is just a circle with lines around it. We know how to make lines! We can just make a series of line plots. Similarly for the central circle, we can make a scatterplot of a single point.

For the moon, how do we make a crescent? Let's be creative and draw a circle as a base (again a scatterplot), with an overlapping circle that is offset and the same colour as the background. The result will be a crescent.

```python
fig, ax = plt.subplots(figsize=(15.66/1.5,9/1.5))

for index, row in df.iterrows():

    ax.bar(row['x_pos'], width=row['x_width'], 
           bottom=row['start'], height=row['lower'] - row['start'],  
           color="#051b2c", edgecolor="#4789be")
    
    ax.bar(row['x_pos'], width=row['x_width'], 
           bottom=row['lower'], height=row['upper'] - row['lower'],
           hatch="///////", color="#4789be")
    

# Step 2
fig.patch.set_facecolor("#0c3759") 
ax.set_facecolor("#4789be")

ax.set_xlim((0, 75))
ax.set_ylim((0, 24))

ax.set_ylabel('Hours\nof sleep', rotation='horizontal', color='#f3e2c0', fontweight='bold')
ax.yaxis.set_label_coords(-0.07, 0.58)

ax.set_xticks([1, 6, 12, 18, 65])
ax.set_xticklabels(['1', '6', '12', '18', ''])
ax.set_yticks(range(0, 25))
ax.set_yticklabels([0] + ['']*5 + [6] + [''] + [8] + [''] + [10] + [''] + [12] + ['']*5 + [18] + ['']*5 + [24])
ax.tick_params('y', color='#f3e2c0', labelcolor='#f3e2c0')
ax.tick_params('x', length=0, labelcolor='#f3e2c0')

ax.text(-1.5, -1.6, 'year old', color='#f3e2c0')
ax.text(65, -0.85, '65 and older', color='#f3e2c0')

ax.spines['left'].set_position(('outward', 8))
ax.spines['left'].set_visible(False)

# Step 3
ax.plot([18, 18], [0, 24], color='white', linestyle='--', linewidth=1)

ax.text(39.5, 16, 'AWAKE', weight='normal', color='white', fontsize=14)
ax.text(39.2, 2.5, 'ASLEEP', weight='normal', color='white', fontsize=14)

ax.text(37, 7.5, 'Seven or more hours', weight='normal', color='white', fontsize=9)
ax.text(1.1, 14.3, '11-14', weight='normal', color='white', fontsize=9)
ax.text(3.1, 13.3, '10-13', weight='normal', color='white', fontsize=9)
ax.text(7.6, 12.3, '9-12', weight='normal', color='white', fontsize=9)
ax.text(13.7, 10.3, '8-10', weight='normal', color='white', fontsize=9)
ax.text(41.9, 9.3, '7-9', weight='normal', color='white', fontsize=9)
ax.text(69, 8.3, '7-8', weight='normal', color='white', fontsize=9)

ax.text(6, 23, 'Sleep per 24 hours', weight='normal', color='white', fontsize=9)
ax.text(8.4, 22.2, 'including naps', weight='normal', color='white', fontsize=9)

ax.text(19, 23, 'Per night', weight='normal', color='white', fontsize=9)

ax.text(1.8, 19, 'Newborn', weight='bold', color='white', fontsize=9)
ax.text(1.8, 18.2, '14-17 hours', weight='normal', color='white', fontsize=9)
ax.text(1.8, 16.5, 'Infant', weight='bold', color='white', fontsize=9)
ax.text(1.8, 15.7, '12-16', weight='normal', color='white', fontsize=9)


# Step 4

# Moon
ax.scatter(42.5, 5, marker='o', s=400, color='#6ebeec')
ax.scatter(43, 5.3, marker='o', s=400, color='#051b2c')

# # Sun
ax.scatter(42.5, 18.5, marker='o', s=150, color='#95cef2')
radius1 = 1.1
radius2 = 1.6
for i in range(24):

    x = [radius1 * math.cos(i / 24 * 2 * 3.14159) + 42.5,
         radius2 * math.cos(i / 24 * 2 * 3.14159) + 42.5]
    y = [radius1 * math.sin(i / 24 * 2 * 3.14159) * 9/16 + 18.5,
         radius2 * math.sin(i / 24 * 2 * 3.14159) * 9/16 + 18.5]
    
    ax.plot(x, y, color='#95cef2', linewidth=1)
    
```

# Visualization Step 5 - Age Arrows

The x-axis label has Age with arrows pointing to the left and right. How can we make that?

Well, we know how to make lines! We can use a lineplot. However, Matplotlib will not draw lineplots outside of the frame of the axis. That area below the x-axis is not an area we can draw lineplots. What can we do?

Let's circumvent this problem by making a new axis in that area of the figure. We can then draw lines on that axis.

```python
fig, ax = plt.subplots(figsize=(15.66/1.5,9/1.5))

for index, row in df.iterrows():

    ax.bar(row['x_pos'], width=row['x_width'], 
           bottom=row['start'], height=row['lower'] - row['start'],  
           color="#051b2c", edgecolor="#4789be")
    
    ax.bar(row['x_pos'], width=row['x_width'], 
           bottom=row['lower'], height=row['upper'] - row['lower'],
           hatch="///////", color="#4789be")
    

# Step 2
fig.patch.set_facecolor("#0c3759") 
ax.set_facecolor("#4789be")

ax.set_xlim((0, 75))
ax.set_ylim((0, 24))

ax.set_ylabel('Hours\nof sleep', rotation='horizontal', color='#f3e2c0', fontweight='bold')
ax.yaxis.set_label_coords(-0.07, 0.58)

ax.set_xticks([1, 6, 12, 18, 65])
ax.set_xticklabels(['1', '6', '12', '18', ''])
ax.set_yticks(range(0, 25))
ax.set_yticklabels([0] + ['']*5 + [6] + [''] + [8] + [''] + [10] + [''] + [12] + ['']*5 + [18] + ['']*5 + [24])
ax.tick_params('y', color='#f3e2c0', labelcolor='#f3e2c0')
ax.tick_params('x', length=0, labelcolor='#f3e2c0')

ax.text(-1.5, -1.6, 'year old', color='#f3e2c0')
ax.text(65, -0.85, '65 and older', color='#f3e2c0')

ax.spines['left'].set_position(('outward', 8))
ax.spines['left'].set_visible(False)

# Step 3
ax.plot([18, 18], [0, 24], color='white', linestyle='--', linewidth=1)

ax.text(39.5, 16, 'AWAKE', weight='normal', color='white', fontsize=14)
ax.text(39.2, 2.5, 'ASLEEP', weight='normal', color='white', fontsize=14)

ax.text(37, 7.5, 'Seven or more hours', weight='normal', color='white', fontsize=9)
ax.text(1.1, 14.3, '11-14', weight='normal', color='white', fontsize=9)
ax.text(3.1, 13.3, '10-13', weight='normal', color='white', fontsize=9)
ax.text(7.6, 12.3, '9-12', weight='normal', color='white', fontsize=9)
ax.text(13.7, 10.3, '8-10', weight='normal', color='white', fontsize=9)
ax.text(41.9, 9.3, '7-9', weight='normal', color='white', fontsize=9)
ax.text(69, 8.3, '7-8', weight='normal', color='white', fontsize=9)

ax.text(6, 23, 'Sleep per 24 hours', weight='normal', color='white', fontsize=9)
ax.text(8.4, 22.2, 'including naps', weight='normal', color='white', fontsize=9)

ax.text(19, 23, 'Per night', weight='normal', color='white', fontsize=9)

ax.text(1.8, 19, 'Newborn', weight='bold', color='white', fontsize=9)
ax.text(1.8, 18.2, '14-17 hours', weight='normal', color='white', fontsize=9)
ax.text(1.8, 16.5, 'Infant', weight='bold', color='white', fontsize=9)
ax.text(1.8, 15.7, '12-16', weight='normal', color='white', fontsize=9)


# Step 4
ax.scatter(42.5, 5, marker='o', s=400, color='#6ebeec')
ax.scatter(43, 5.3, marker='o', s=400, color='#051b2c')

ax.scatter(42.5, 18.5, marker='o', s=150, color='#95cef2')
radius1 = 1.1
radius2 = 1.6
for i in range(24):

    x = [radius1 * math.cos(i / 24 * 2 * 3.14159) + 42.5,
         radius2 * math.cos(i / 24 * 2 * 3.14159) + 42.5]
    y = [radius1 * math.sin(i / 24 * 2 * 3.14159) * 9/16 + 18.5,
         radius2 * math.sin(i / 24 * 2 * 3.14159) * 9/16 + 18.5]
    
    ax.plot(x, y, color='#95cef2', linewidth=1)


# Step 5

# Newborn / infant lines
ax.plot([0.3, 0.4, 0.65, 0.9, 1.1, 1.3, 1.5], [17.4, 17.9, 18.2, 18.35, 18.44, 18.47, 18.5], color='white', linewidth=0.4)
ax.plot([0.7, 0.77, 0.9, 1.05, 1.2, 1.3, 1.5], [15.4, 15.8, 16.05, 16.17, 16.22, 16.26, 16.3], color='white', linewidth=0.4)
    
# Age arrows
ax_age = fig.add_axes((0.4, 0, 0.32, 0.1))

ax_age.set_facecolor("#0c3759")

for side in ['left', 'bottom', 'right', 'top']:
    ax_age.spines[side].set_visible(False)
    
ax_age.set_xlim((0, 1))
ax_age.set_ylim((0, 1))
ax_age.set_xticks([])
ax_age.set_yticks([])

ax_age.plot([0.2, 0.44], [0.45, 0.45], color='#f3e2c0', linewidth=0.5)
ax_age.plot([0.595, 0.835], [0.45, 0.45], color='#f3e2c0', linewidth=0.5)

ax_age.scatter(0.2, 0.45, color='#f3e2c0', linewidth=0.5, marker='<', s=8)
ax_age.scatter(0.835, 0.45, color='#f3e2c0', linewidth=0.5, marker='>', s=8)

ax_age.text(0.48, 0.39, 'Age', weight='bold', color='#f3e2c0')
```
