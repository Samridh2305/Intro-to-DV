```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("river-plastics.csv")
fig, ax = plt.subplots(figsize=(10, 5))


ax.spines["left"].set_visible(True)
ax.spines["bottom"].set_visible(True)
ax.spines["right"].set_visible(False)
ax.spines["top"].set_visible(False)

x = df["Coast length"] / df["Area"]

y = (df["ME"] / df["MPW"]) * 100

scatterplot = ax.scatter(
    x,
    y,
    c=df["Rainfall"],
    cmap="viridis"
)

ax.set_xscale("log")
ax.set_yscale("log")

ax.grid(True, which="major", linewidth=0.4, alpha=0.4)

ax.set_xlabel("Ratio of coastal length to area")
ax.set_ylabel("Plastic waste in oceans (% of total plastic waste)")
ax.set_title("Drivers of Plastic Pollution in the Oceans")

cbar = fig.colorbar(scatterplot, ax=ax)
cbar.set_label("Rainfall [mm / year]")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sleep.csv")

df = df.iloc[::-1].reset_index(drop=True)

y = range(len(df))

lower_color = "#051B2C"
upper_color = "#4789BE"

fig, ax = plt.subplots(figsize=(12, 5))

ax.set_axisbelow(True)

for i, row in df.iterrows():
    ax.plot(
        [row["Recommended lower"], row["Recommended upper"]],
        [i, i],
        color=lower_color,
        linewidth=1.75,
        zorder=1
    )

ax.scatter(
    df["Recommended lower"],
    y,
    color=lower_color,
    s=28,
    zorder=2
)

ax.scatter(
    df["Recommended upper"],
    y,
    color=upper_color,
    s=28,
    zorder=2
)

ax.set_yticks(list(y))
ax.set_yticklabels([
    "Older adults",
    "Adults",
    "Young adults",
    "Teenagers",
    "School-aged children",
    "Preschoolers",
    "Toddlers",
    "Infants",
    "Newborns"
])

ax.set_xlim(0, 24)
ax.set_xticks([0, 6, 8, 10, 12, 14, 18, 24])

ax.xaxis.tick_top()
ax.tick_params(
    axis="x",
    which="both",
    top=True,
    bottom=False,
    labeltop=True,
    labelbottom=False,
    length=3
)

ax.spines["left"].set_visible(False)
ax.spines["bottom"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["top"].set_visible(False)

ax.tick_params(axis="y", length=0)

ax.grid(
    axis="x",
    color="#eeeeee",
    linewidth=0.8
)

ax.grid(
    axis="y",
    color="#eeeeee",
    linewidth=0.8
)

ax.set_title(
    "National Sleep Foundation's recommended hours of sleep",
    fontsize=14,
    pad=15
)

ax.tick_params(axis="both", labelsize=9)

ax.tick_params(axis="y", pad=5)

plt.tight_layout()
plt.show()
```

```python
The recommended amount of sleep decreases as we grow up. 14 to 17 hours of sleep per day is recommended for Newborns, 
while infants have a slightly less 12 to 15 hours recommended sleep time and toddlers need even less sleep time of 11 to 14 hours.
The recommended range continues to decrease from 10 to 14 hours
for preschoolers to 9 to 11 hours for school-aged children, and 8 to 10 for teenagers respectively. 
Young adults and adults have the same recommended range of 7 to 9 hours, 
while older adults have the lease recommended sleep range of 7 to 8 hours.

A connected dot plot shows both the lower and upper recommended values within a line while connecting 
them and thus showcasing the range of values recommended.
The length of each connecting line makes the size of the recommended sleep range easy to compare across age groups.
If we draw bar graphs for specific age groups it makes it difficult for the viewer to compare bar heights for each individual age, 
whereas the connected dots emphasize the range between the two recommended values.


```
