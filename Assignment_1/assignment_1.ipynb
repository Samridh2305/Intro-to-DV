## Question 1 - Scatter Plot - Waste by Country (based on coastal line, area and rainfall)

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("river-plastics.csv")
fig, ax = plt.subplots(figsize=(10, 6))


ax.spines["left"].set_visible(True)
ax.spines["bottom"].set_visible(True)
ax.spines["right"].set_visible(False)
ax.spines["top"].set_visible(False)

ax.spines["left"].set_color("#d0d0d0")
ax.spines["bottom"].set_color("#d0d0d0")
ax.spines["left"].set_linewidth(1)
ax.spines["bottom"].set_linewidth(1)

x = data["Coast length"] / data["Area"]

y = (data["ME"] / data["MPW"]) * 100

# vmin=0 makes the colour bar start at 0 mm/year, as in the reference figure
scatterplot = ax.scatter(
    x,
    y,
    c=data["Rainfall"],
    cmap="viridis",
    vmin=0
)

ax.set_xscale("log")
ax.set_yscale("log")

# Same y-range as the reference figure. Jordan (0.0008%) falls below it and is not shown there either.
ax.set_ylim(0.008, 30)

ax.grid(True, which="major", color="#f0f0f0", linewidth=0.8)

ax.set_xlabel("Ratio of coastal length to area", fontsize=12)
ax.set_ylabel("Plastic waste in oceans (% of total plastic waste)", fontsize=12)
ax.set_title("Drivers of Plastic Pollution in the Oceans", fontsize=14)

cbar = fig.colorbar(scatterplot, ax=ax)
cbar.set_label("Rainfall [mm / year]", fontsize=12)
cbar.outline.set_edgecolor("#d0d0d0")
cbar.outline.set_linewidth(1)

plt.tight_layout(pad=1.2)
plt.show()
```

## Question 2 - Connected Dot Plot - How much should you be sleeping?

```python
import pandas as pd
import matplotlib.pyplot as plt

sleep = pd.read_csv("sleep.csv")

sleep = sleep.iloc[::-1].reset_index(drop=True)

y = range(len(sleep))

lower_color = "#051B2C"
upper_color = "#4789BE"

fig, ax = plt.subplots(figsize=(10, 6))

ax.set_axisbelow(True)

for i, row in sleep.iterrows():
    ax.plot(
        [row["Recommended lower"], row["Recommended upper"]],
        [i, i],
        color="black",
        linewidth=1.5,
        zorder=1
    )

ax.scatter(
    sleep["Recommended lower"],
    y,
    color=lower_color,
    edgecolors="black",
    linewidths=1,
    zorder=2
)

ax.scatter(
    sleep["Recommended upper"],
    y,
    color=upper_color,
    edgecolors="black",
    linewidths=1,
    zorder=2
)

ax.set_yticks(list(y))
ax.set_yticklabels(sleep["Age group"])

ax.set_xlim(0, 24)
ax.set_xticks([0, 6, 8, 10, 12, 14, 18, 24])

ax.xaxis.tick_top()
ax.tick_params(
    axis="x",
    which="both",
    top=True,
    bottom=False,
    labeltop=True,
    labelbottom=False
)

ax.spines["left"].set_visible(False)
ax.spines["bottom"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["top"].set_visible(False)

ax.tick_params(axis="y", length=0)

ax.grid(
    axis="x",
    color="#f0f0f0",
    linewidth=0.8
)

ax.grid(
    axis="y",
    color="#f0f0f0",
    linewidth=0.8
)

ax.set_title(
    "National Sleep Foundation's recommended hours of sleep",
    fontsize=14,
    pad=18
)

plt.tight_layout(pad=1.2)
plt.show()
```

**Question 2 written response:** Describe how the recommended sleep range changes across the lifespan, and why a connected dot plot communicates this better than two bars per group.

The recommended amount of sleep decreases as we grow up. 14 to 17 hours of sleep per day is recommended for Newborns, 
while infants have a slightly less 12 to 15 hours recommended sleep time and toddlers need even less sleep time of 11 to 14 hours.
The recommended range continues to decrease from 10 to 13 hours
for preschoolers to 9 to 12 hours for school-aged children, and 8 to 10 for teenagers respectively. 
Young adults and adults have the same recommended range of 7 to 9 hours, 
while older adults have the least recommended sleep range of 7 to 8 hours.

A connected dot plot shows both the lower and upper recommended values within a line while connecting 
them and thus showcasing the range of values recommended.
The length of each connecting line makes the size of the recommended sleep range easy to compare across age groups.
If we draw bar graphs for specific age groups it makes it difficult for the viewer to compare bar heights for each individual age, 
whereas the connected dots emphasize the range between the two recommended values.
